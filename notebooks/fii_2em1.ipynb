```python
import requests
import pandas as pd
import io

# 1. Configuração da URL e Headers (para imitar um navegador)
url = "https://www.fundamentus.com.br/fii_resultado.php"

# É importante definir o User-Agent para o site não bloquear o script achando que é um robô malicioso
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36"
}

# 2. Fazer a requisição POST (conforme mostrado no seu DevTools)
print("1. Realizando a requisição POST...")
response = requests.post(url, headers=headers)

# Verificação se deu certo (Status 200 OK)
if response.status_code == 200:
    print("   Sucesso! Conteúdo baixado.")

    # Importante: A imagem mostra que o charset é iso-8859-1.
    # Vamos forçar essa codificação para não ter problemas com acentos.
    response.encoding = 'iso-8859-1'

    # 3. Extrair a tabela do HTML usando Pandas
    # O read_html procura por tags <table> e retorna uma lista de DataFrames
    # O parâmetro decimal e thousands ajudam o pandas a entender os números no formato BR
    print("2. Extraindo a tabela...")
    try:
        tabelas = pd.read_html(io.StringIO(response.text), decimal=',', thousands='.')

        # Normalmente a tabela principal é a primeira encontrada (índice 0)
        df = tabelas[0]

        # Visualizar as primeiras linhas para conferência
        print(f"   Tabela encontrada com {len(df)} linhas.")
        # print(df.head())

        # 4. Transformar em CSV
        print("3. Salvando em CSV...")
        # Usamos sep=';' porque no Brasil o Excel usa vírgula para decimais
        # encoding='utf-8-sig' garante que acentos apareçam corretamente no Excel
        df.to_csv('/content/drive/MyDrive/Colab Notebooks/Investimentos/fii_resultado.csv', index=False, sep=',', encoding='utf-8-sig')

        print("   Arquivo 'fii_resultado.csv' criado com sucesso!")

    except Exception as e:
        print(f"   Erro ao processar a tabela: {e}")
else:
    print(f"Erro na requisição. Status Code: {response.status_code}")
```

## Aplicando filtros


*   Liquidez > 200.000
*   Dividend Yield > 7
*   P/VP entre 0,69 e 1,05



```python
LIQUIDEZ_MIN = 200000
DIVIDEND_YIELD_MIN = 7
P_VP_MIN = 0.69
P_VP_MAX = 1.05
```

```python
# Filtra Liquidez > 400000
fii_aux = fii_aux[fii_aux['Liquidez'] > LIQUIDEZ_MIN]

# Transforma coluna Dividend Yield de fii_aux em float
fii_aux['Dividend Yield'] = (
    fii_aux['Dividend Yield']
    .astype(str)
    .str.replace('%', '', regex=False)
    .str.replace(',', '.', regex=False)
    .astype(float)
)

# Filtra Dividend Yield > 7%
fii_aux = fii_aux[fii_aux['Dividend Yield'] > DIVIDEND_YIELD_MIN]

# Filtrar 0.69 > P/VP < 1.05
fii_aux = fii_aux[fii_aux['P/VP'] > P_VP_MIN]
fii_aux = fii_aux[fii_aux['P/VP'] < P_VP_MAX]
```

## Aplicando o Método 2 em 1
* Rank DY + Rank PVP = Rank 2 em 1

```python
# Ordena a tabela por dividend Yield
fii_aux = fii_aux.sort_values(by='Dividend Yield', ascending=False)

# Cria uma coluna de Rank inteiro
fii_aux['Rank DY'] = range(1, len(fii_aux) + 1)

# Ordena a tabela por P/VP
fii_aux = fii_aux.sort_values(by='P/VP', ascending=True)

# Cria uma coluna de Rank inteiro
fii_aux['Rank PVP'] = range(1, len(fii_aux) + 1)

# Cria a coluna 2em1 que é a soma dos ranks DY e PVP
fii_aux['2em1'] = fii_aux['Rank DY'] + fii_aux['Rank PVP']

fii_aux.to_csv('/content/drive/MyDrive/Colab Notebooks/Investimentos/fii_2em1.csv', index=False, sep=',', encoding='utf-8-sig')
```

## Visualizando por Segmentos



```python
def class_por_segmento(segmento):
    # Cria uma tabela só com Segmento Shoppings
    fii = fii_aux[fii_aux['Segmento'] == segmento]

    # Ordena a tabela por 2em1
    fii = fii.sort_values(by='2em1', ascending=True)
    return fii
```
